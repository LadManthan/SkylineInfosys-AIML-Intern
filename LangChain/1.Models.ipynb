## using t5 base model

```python
from langchain_huggingface import HuggingFacePipeline
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM, pipeline

model_id = "google/flan-t5-base"
tokenizer = AutoTokenizer.from_pretrained(model_id)
model = AutoModelForSeq2SeqLM.from_pretrained(model_id)

pipe = pipeline(
    "text2text-generation",
    model=model,
    tokenizer=tokenizer,
    max_new_tokens=50
)

llm = HuggingFacePipeline(pipeline=pipe)
```

```python
result = llm.invoke("how many continents are there in the world?")
print(result)

result = llm.invoke("how to make a cake?")
print(result)
```

## using t5 large model

```python
model_id = "google/flan-t5-large"
tokenizer = AutoTokenizer.from_pretrained(model_id)
model = AutoModelForSeq2SeqLM.from_pretrained(model_id)

pipe = pipeline(
    "text2text-generation",
    model=model,
    tokenizer=tokenizer,
    max_new_tokens=100
)

llm = HuggingFacePipeline(pipeline=pipe)
```

```python
que = [
    'Who was the first presidents of USA?',
    'what is the process of creating a bank account?',
    'what is the color of the sun and moon?',
    'what will happen if we dont breath for 10 minutes?',
    'what are the tallest mountain ranges?'
]

for q in que:
  result = llm.invoke(q)
  print(f"\n{q} : {result}")
```

## embedding model

```python
from langchain_huggingface import HuggingFaceEmbeddings

embedding = HuggingFaceEmbeddings(model_name='sentence-transformers/all-MiniLM-L6-v2')
```

```python
text = "Heyy there! I am Manthan."
vector = embedding.embed_query(text)
print(vector)
```

```python
document = [
    'Planets revolve around the sun.',
    'earth has only one moon',
    'there are 8 planets in the continent',
    'okay byee'
]

vector = embedding.embed_documents(document)

for v in vector:
  print(v)
```

## simple qa bot

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
doc = [
    "Virat Kohli is a great batter who has scored many centuries for India.",
    "Rohit Sharma is known for his powerful hitting and leading India to victory.",
    "MS Dhoni is a famous captain who won the World Cup for India in 2011.",
    "Sachin Tendulkar is called the God of Cricket and has the most runs in history.",
    "Jasprit Bumrah is a very fast bowler who takes many wickets with his yorkers.",
    "Kapil Dev was the captain who won India's first World Cup in 1983.",
]

doc_embedding = embedding.embed_documents(doc)
```

```python
while True:
  query = input("\nEnter your query: ")
  if query == 'exit':
    break
  else:
    query_embeddings = embedding.embed_query(query)

    scores = cosine_similarity([query_embeddings],doc_embedding)[0]

    index,score = sorted(list(enumerate(scores)),key=lambda x:x[1])[-1]

    print(doc[index])
    print(f"Score : {score}")
```
