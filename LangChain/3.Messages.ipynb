```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM, pipeline
from langchain_huggingface import HuggingFacePipeline
```

```python
model_id = "google/flan-t5-base"

tokenizer = AutoTokenizer.from_pretrained(model_id)
model = AutoModelForSeq2SeqLM.from_pretrained(model_id)

pipe = pipeline(
    "text2text-generation",
    model=model,
    tokenizer=tokenizer,
    max_new_tokens=100
)

llm = HuggingFacePipeline(pipeline=pipe)
```

## Chat History

```python
chat_history = []

while True:
    user_input = input('\nYou : ')
    chat_history.append(user_input)
    if user_input == 'exit':
        break
    result = llm.invoke(chat_history)
    chat_history.append(result)
    print(result)

print('-'*80,"\nChat History\n",'-'*80)
print(chat_history)
```

## Messages

**SystemMessage** — Defines the assistant’s role, behavior, or instructions for the conversation.  
**HumanMessage** — Represents the user’s input or question in the chat.  
**AIMessage** — Represents the model’s response to the user.  

```python
from langchain_core.messages import SystemMessage, HumanMessage, AIMessage
```

```python
model_id = "google/flan-t5-large"

tokenizer = AutoTokenizer.from_pretrained(model_id)
model = AutoModelForSeq2SeqLM.from_pretrained(model_id)

pipe = pipeline(
    "text2text-generation",
    model=model,
    tokenizer=tokenizer,
    max_new_tokens=100
)

llm = HuggingFacePipeline(pipeline=pipe)
```

```python
chat_history = [
    SystemMessage(content='You are a helpful assistance')
]

while True:
    user_input = input('\nYou : ')
    chat_history.append(HumanMessage(content=user_input))
    
    if user_input == 'exit':
        break

    result = llm.invoke(chat_history)
    chat_history.append(AIMessage(content=result))
    print(result)

print('-'*80,"\nChat History\n",'-'*80)
print(chat_history)
```

## Chat Prompt
#### It lets us to create a structured conversation prompt with roles (system and human)

```python
from langchain_core.prompts import ChatPromptTemplate
from transformers import AutoModelForCausalLM
import torch
```

```python
model_id = "TinyLlama/TinyLlama-1.1B-Chat-v1.0"
tokenizer = AutoTokenizer.from_pretrained(model_id)
model = AutoModelForCausalLM.from_pretrained(
    model_id,
    dtype=torch.float16,
    device_map='auto'
)

pipe = pipeline(
    'text-generation',
    model=model,
    tokenizer = tokenizer,
    max_new_tokens = 100,
)

cricket_llm = HuggingFacePipeline(pipeline=pipe)
```

```python
chat_template = ChatPromptTemplate([
    ('system','You are a helpful {domain} expert'),
    ('human','Explain in simple terms, what is {topic}')
])

prompt = chat_template.invoke({'domain':'cricket','topic':'cricket'})
print(prompt)
```

```python
print(prompt.to_string())
```

```python
response = cricket_llm.invoke(prompt.to_string())
print(response)
```

## Message Placeholder
#### It is a slot in a chat prompt template where a list of messages — usually conversation history — gets inserted dynamically.

```python
from langchain_core.prompts import MessagesPlaceholder
```

```python
#create a chat template
chat_template = ChatPromptTemplate([
    ("system", "You are a helpful assistant."),
    MessagesPlaceholder(variable_name="chat_history"),
    ("human", "{question}")
])
```

```python
#loading chat history
chat_history = []

with open('/kaggle/input/langchain/chat_history.txt') as f:
    chat_history.extend(f.readlines())

print(chat_history)
```

```python
prompt = chat_template.invoke({
    "chat_history": chat_history,
    "question": "How to make a omlet?"
})

print(prompt)
```

```python
print(prompt.to_string())
```

```python
while True:
    user_input = input("You: ")

    if user_input.lower() == "exit":
        break

    # Create prompt
    prompt = chat_template.invoke({
        "chat_history": chat_history,
        "question": user_input
    })

    # Get model response
    response = llm.invoke(prompt.to_string())

    print("Bot:", response)

    # Update memory automatically
    chat_history.append(HumanMessage(content=user_input))
    chat_history.append(AIMessage(content=response))
```
