# Text Loader

```python
from langchain_groq import ChatGroq
from langchain_community.document_loaders import TextLoader
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

```python
model = ChatGroq(model="groq/compound")

prompt = PromptTemplate(
    template='List out the key points from the following text : {text}',
    input_variables=['text']
)

parser = StrOutputParser()
```

```python
loader = TextLoader('/kaggle/input/rag-document-loader/cricket.txt')

docs = loader.load()
print(type(docs))
```

```python
print(len(docs))
```

```python
print(docs[0])
```

#### page content and meta deta

```python
print(docs[0].page_content)
```

```python
print(docs[0].metadata)
```

```python
chain = prompt | model | parser
print(chain.invoke({'text':docs[0].page_content}))
```

# Directory Loader
Load multiple documents from a folder at once instead of loading files one by one.  
Scans a directory  
Reads all matching files (e.g. .txt, .pdf, .md)  
Converts them into Document objects  

**Loader vs Lazy Loader**  
**Loader**:  
It loads all documents at once, Stores everything in memory  
High memory usage & Slow for large datasets  

**Lazy Loader**:  
Loads documents one by one, Uses a generator (streaming)  
memory efficient but complex

# CSV Loader

```python
from langchain_community.document_loaders import CSVLoader
```

```python
loader = CSVLoader(file_path='/kaggle/input/rag-document-loader/data.csv')
data = loader.load()
print(len(data))  #no. of rows
```

```python
print(data[0].page_content)  #1st row
```

```python
print(data[0].metadata)
```

```python
prompt = PromptTemplate(
    template='Return the most sold product and its revenue from this file : {file}',
    input_variables=['file']
)

parser = StrOutputParser()

chain = prompt | model | parser

all_rows = "\n".join(doc.page_content for doc in data)

result = chain.invoke({
    "file": all_rows
})

print(result)
```
