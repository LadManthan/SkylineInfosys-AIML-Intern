# 1.Length Based

```python
from langchain_text_splitters import CharacterTextSplitter
```

```python
text = "Cricket is a bat-and-ball game that is played between two teams of eleven players on a field, at the centre of which is a 22-yard (20-metre; 66-foot) pitch with a wicket at each end, each comprising two bails (small sticks) balanced on three stumps. Two players from the batting team, the striker and nonstriker, stand in front of either wicket holding bats, while one player from the fielding team, the bowler, bowls the ball toward the striker's wicket from the opposite end of the pitch. The striker's goal is to hit the bowled ball with the bat and then switch places with the nonstriker, with the batting team scoring one run for each of these swaps. Runs are also scored when the ball reaches the boundary of the field or when the ball is bowled illegally."
print(text)
```

```python
splitter = CharacterTextSplitter(
    chunk_size=150,
    chunk_overlap=0, #overlap characters between 2 chunks
    separator=''
)

result = splitter.split_text(text)

print(f"Total number of chunks : {len(result)}\n")

for res in result:
    print(f"\n{res} | characters : {len(res)}")
```

#### using chunk_overlap

```python
splitter = CharacterTextSplitter(
    chunk_size=150,
    chunk_overlap=10, #overlap characters between 2 chunks
    separator=''
)

result = splitter.split_text(text)

print(f"Total number of chunks : {len(result)}\n")

for res in result:
    print(f"\n{res} | characters : {len(res)}")
```

# 2.Text Structure

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
text = """
Space exploration has led to incredible scientific discoveries. From landing on the Moon to exploring Mars, humanity continues to push the boundaries of what’s possible beyond our planet.

These missions have not only expanded our knowledge of the universe but have also contributed to advancements in technology here on Earth. Satellite communications, GPS, and even certain medical imaging techniques trace their roots back to innovations driven by space programs.
"""

splitter = RecursiveCharacterTextSplitter(
    chunk_size=300,
    chunk_overlap=0,
)

chunks = splitter.split_text(text)
print(f"Total chunks : {len(chunks)}")

for chunk in chunks:
    print(f"\n{'-'*60}\n characters : {len(chunk)}\n{'-'*60}")
    print(f"{chunk}")
```

# 3.Document Structure

## Code based

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter, Language
```

```python
code = """
class Car:
    def __init__(self, brand, model, mileage):
        self.brand = brand
        self.model = model
        self.mileage = mileage  # mileage in km/l (float)

    def get_details(self):
        return f"{self.brand} {self.model}"

    def is_fuel_efficient(self):
        return self.mileage >= 15.0


# creating object
car1 = Car("Toyota", "Corolla", 18.5)
print(car1.get_details())

if car1.is_fuel_efficient():
    print("The car is fuel efficient.")
else:
    print("The car is not fuel efficient.")
"""
```

```python
splitter = RecursiveCharacterTextSplitter.from_language(
    language=Language.PYTHON,
    chunk_size=200,
    chunk_overlap=0
)

chunks = splitter.split_text(code)
print(f"Total chunks : {len(chunks)}")

for chunk in chunks:
    print(f"\n{'-'*60}\n characters : {len(chunk)}\n{'-'*60}")
    print(f"{chunk}")
```

# 4.Semantic Meaning Based

```python
from langchain_experimental.text_splitter import SemanticChunker
from langchain_community.embeddings import HuggingFaceEmbeddings
```

```python
text = """
Cricket is a popular sport played between two teams where players score runs by hitting the ball with a bat. It is especially loved in countries like India, Australia, New Zeland, South Africa and England.

Global warming refers to the long-term rise in Earth’s average temperature due to increased greenhouse gas emissions. It leads to climate changes such as melting glaciers, rising sea levels, and extreme weather events.
"""
```

```python
embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)
```

```python
splitter = SemanticChunker(
    embeddings,
    breakpoint_threshold_type="standard_deviation",
    breakpoint_threshold_amount=1
)

chunks = splitter.split_text(text)
print(f"Total chunks : {len(chunks)}")

for chunk in chunks:
    print(f"\n{'-'*60}\n characters : {len(chunk)}\n{'-'*60}")
    print(f"{chunk}")
```
